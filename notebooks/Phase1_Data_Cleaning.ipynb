# Phase 1: Data Cleaning & Normalization Notebook

Welcome! In this notebook, we will interactively load, inspect, clean brand typos, and export the cleaned dataset step-by-step using **Pandas**.

## Cell 1: Import Libraries & Configure File Paths

```python
import os
import pandas as pd
from pathlib import Path

# Define project data directory paths
DATA_DIR = Path("data")
INPUT_CSV = DATA_DIR / "multinational_brand_shortlist.csv"
OUTPUT_CSV = DATA_DIR / "multinational_brand_cleaned.csv"

print(f"Input CSV Path: {INPUT_CSV.resolve()}")
```

## Cell 2: Load the Raw Dataset & Inspect

```python
# Load raw CSV file into a Pandas DataFrame
df = pd.read_csv(INPUT_CSV)

print("=== Dataset Overview ===")
print(f"Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
df.head(10)
```

```python
# Loop through the array and print each item on a new line
for item in df['Item name'].unique():
    print(item)
```

```python
# since cadbury and bournvita and highly confusing lets see what type of products are they??
# Search for brand names containing "CADB" OR "BOURN" OR "BOUENV"
cadbury_bournvita_rows = df[
    df['Brand_Name'].str.contains('CADB|BOURN|BOUENV', case=False, na=False)
]

# View specific columns
cadbury_bournvita_rows[['Brand_Name', 'Item name', 'Category']]
```

```python
# 1. The Updated Typo Dictionary 
typo_map = {
    'HERSHYEYS': 'HERSHEYS',
    'BOUENVITA': 'CADBURY',
    'CADBERY BORNVITA': 'CADBURY',
    'BOURNVILLE': 'CADBURY', 
    'MOUMTAIN DEW': 'MOUNTAIN DEW',
    'DR OETKERS FUNFOODS': 'DR OETKER'
}

# 2. Clean the Brand_Name column in place
df['Brand_Name'] = df['Brand_Name'].replace(typo_map)

# 3. Save the cleaned dataframe to a new CSV file
df.to_csv(OUTPUT_CSV, index=False)

# 4. Verify that the typos are gone
print("Cleaned unique brands list:\n", df['Brand_Name'].unique())
print("\nTotal unique brands now:", df['Brand_Name'].nunique())
```

# VERIFYING THE NEW DATA


```python
df_cleaned = pd.read_csv(OUTPUT_CSV)
print("Cleaned Brands list:")
print(df_cleaned['Brand_Name'].unique())
```

```python
# 3. Verify Cadbury absorbed everything (Bournvita, Bournville, etc.)
cadbury_rows = df_cleaned[df_cleaned['Brand_Name'] == 'CADBURY']
print(f"\nTotal CADBURY products after merge: {len(cadbury_rows)}")
```

```python
# Load raw vs cleaned to compare side-by-side
df_raw = pd.read_csv("data/multinational_brand_shortlist.csv")
df_clean = pd.read_csv("data/multinational_brand_cleaned.csv")

# Filter rows where the brand name was actually changed
changes = df_raw[df_raw['Brand_Name'] != df_clean['Brand_Name']]

# Display before vs after
print(f"Total Rows Corrected: {len(changes)}")
changes_summary = pd.DataFrame({
    'Original Brand (Raw)': df_raw.loc[changes.index, 'Brand_Name'],
    'Cleaned Brand (New)': df_clean.loc[changes.index, 'Brand_Name'],
    'Product Item Name': df_raw.loc[changes.index, 'Item name']
})

changes_summary.drop_duplicates(subset=['Original Brand (Raw)', 'Cleaned Brand (New)'])
```
